```python
from uncertainties import ufloat
import numpy as np
```

```python
# Wasserwert berechnen
m_k_voll = ufloat(427.650,0.01) * 1e-3
m_k_leer = ufloat(251.08,0.01) * 1e-3

m_W = m_k_voll - m_k_leer
c_W = ufloat(4.186,0.004) * 1000 # Wert aus dem Skript
T_1 = ufloat(56.0,1) # Wasser am Anfang
T_2 = ufloat(24.3,1) # Raumtemperatur
T_bar = ufloat(53.06,1.2) # Aus dem Diagramm bestimmt

W = m_W * c_W * (T_1 - T_bar) / (T_bar - T_2)
print(W)
```

```python

W = ufloat(80,10)
# Wärmekapazitäten berechnen
def c_x(m_k_voll,m_x,T_1,T_2,T_misch):
    return ((m_k_voll - m_k_leer) * c_W + W) * (T_misch - T_2) / (m_x * (T_1 - T_misch))

Masse_Probekörper_Groß = {
    "Alu" : ufloat(147.35,0.01) * 1e-3,
    "Blei" : ufloat(603.230, 0.01) * 1e-3,
    "Graphit" : ufloat(121.99,0.01) * 1e-3 }

T_kochendes_Wasser = ufloat(100.2,1)

T_misch = {
    "Alu" : ufloat(29.3,1),
    "Graphit" : ufloat(33.4,1),
    "Blei" : ufloat(36.1,1) }

T_Wasser_vorher = {
    "Alu" : ufloat(23.6,1),
    "Graphit" : ufloat(29.2,1),
    "Blei" : ufloat(33.2,1) }

Molare_Masse = {
    "Alu" : 26.98,
    "Graphit" : 12.01,
    "Blei" : 207.2
}

M_K_Voll =  {
    "Alu" : ufloat(598.37,0.01) * 1e-3,
    "Graphit" : ufloat(597.82,0.01) * 1e-3,
    "Blei" : ufloat(595.90,0.01) * 1e-3
}

C_X_Lit = {
    "Alu" : 129,
    "Graphit" : 900,
    "Blei" : 709
}

print("Alu:", c_x(ufloat(598.37,0.01) * 1e-3, Masse_Probekörper_Groß["Alu"], T_kochendes_Wasser, T_Wasser_vorher["Alu"], T_misch["Alu"]))

print("Graphit:", c_x(ufloat(597.82,0.01) * 1e-3, Masse_Probekörper_Groß["Graphit"], T_kochendes_Wasser, T_Wasser_vorher["Graphit"], T_misch["Graphit"]))
print("Blei:", c_x(ufloat(595.90,0.01) * 1e-3, Masse_Probekörper_Groß["Blei"], T_kochendes_Wasser, T_Wasser_vorher["Blei"], T_misch["Blei"]))



    
```

```python
print("Alu:", Molare_Masse["Alu"] * c_x(ufloat(598.37,0.01) * 1e-3, Masse_Probekörper_Groß["Alu"], T_kochendes_Wasser, T_Wasser_vorher["Alu"], T_misch["Alu"]))

print("Graphit:", Molare_Masse["Graphit"] * c_x(ufloat(597.82,0.01) * 1e-3, Masse_Probekörper_Groß["Graphit"], T_kochendes_Wasser, T_Wasser_vorher["Graphit"], T_misch["Graphit"]))
print("Blei:", Molare_Masse["Blei"] *  c_x(ufloat(595.90,0.01) * 1e-3, Masse_Probekörper_Groß["Blei"], T_kochendes_Wasser, T_Wasser_vorher["Blei"], T_misch["Blei"]))

def printnum(uf):
    return f'num("{str(uf).replace("/","").replace("e+0","e").replace("(","").replace(")","")}")'

def sigma(m, lit):
    return (abs(m.nominal_value - lit)/m.std_dev).__round__(2)

R = 8.314

c_mol_dp = 3 * R

print("Dulong Petit", c_mol_dp)


def print_typst_table_row(name):
    s = f"[{name}],"
    c = c_x(M_K_Voll[name], Masse_Probekörper_Groß[name], T_kochendes_Wasser, T_Wasser_vorher[name], T_misch[name])
    s += printnum(c) + ","
    s += f"${C_X_Lit[name]}$,"
    s += f"${sigma(c, C_X_Lit[name])}$,"
    s += printnum(c * Molare_Masse[name]) + ","
    s += f"${Molare_Masse[name] * C_X_Lit[name]}$,"
    s += f"${sigma(c * Molare_Masse[name], c_mol_dp)}$,"
    print(s)

print_typst_table_row("Alu")
```
