```python
from uncertainties import ufloat
import numpy as np
from uncertainties import unumpy as unp

def SMD(u1,u2): return abs(u1.nominal_value - u2.nominal_value)/np.sqrt(u1.std_dev**2 + u2.std_dev**2)
```

```python
f_1_gemessen = 10 * ufloat(158,3)
f_2_gemessen = 2.5 * ufloat(40,0.08)

f_I_gemessen = ufloat(1594,5)
f_II_gemessen = ufloat(1394,5)

f_2_rechnen = 1/2 * (-f_II_gemessen + f_I_gemessen)
f_1_rechnen = 1/2 * (f_II_gemessen + f_I_gemessen)

print(f_2_gemessen,f_2_rechnen)
print(f_1_gemessen,f_1_rechnen)
```

```python
# PWM 1

U1_eff_gemessen = ufloat(1.24,0.01)
U1_mitt_gemessen = ufloat(444,3) / 1000

T1 = ufloat(1.03,0.01) / 1000
t1 = ufloat(110,10) * 10**(-6) # Pulsbreite
U_0_1 = ufloat(3.44,0.04) # Pulshöhe

U1_mitt_berechnet = U_0_1 * t1 / T1
print(U1_mitt_gemessen, U1_mitt_berechnet, SMD(U1_mitt_gemessen,U1_mitt_berechnet))
U1_eff_berechnet = U_0_1 * unp.sqrt(t1 / T1)
print(U1_eff_gemessen, U1_eff_berechnet, SMD(U1_eff_gemessen,U1_eff_berechnet))

```

```python
# PWM 2

U2_eff_gemessen = ufloat(3.03,0.01)
U2_mitt_gemessen = ufloat(2.7,0.01)

T2 = ufloat(1.03,0.01) / 1000
t2 = ufloat(860,10) * 10**(-6) # Pulsbreite
U_0_2 = ufloat(3.40,0.04) # Pulshöhe

U2_mitt_berechnet = U_0_2 * t2 / T2
print(U2_mitt_gemessen, U2_mitt_berechnet, SMD(U2_mitt_gemessen,U2_mitt_berechnet))
U2_eff_berechnet = U_0_2 * unp.sqrt(t2 / T2)
print(U2_eff_gemessen, U2_eff_berechnet, SMD(U2_eff_gemessen,U2_eff_berechnet))

```
