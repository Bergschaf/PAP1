```python
m_18 = ufloat(111.2849,0.0001) 
m_angefressen = ufloat(93.0577,0.0001) # TODO

# Zielspannung: 1 A
I = ufloat(1, 0.01) # Ampere
T = ufloat(1800,10)

# 14 bei plus

m_18_danach = ufloat(110.6438,0.0001)
m_angefressen_danach = ufloat(93.6621, 0.0001)

delta_m_anode = m_18_danach - m_18
delta_m_kathode = m_angefressen_danach - m_angefressen


Q = I * T


```

```python
M_Mol = 63.546 # Molare Masse von Kupfer in Gramm
z = 2


F_anode = Q/(z * abs(delta_m_anode))  * M_Mol
F_kathode = Q/(z * abs(delta_m_kathode))  * M_Mol

F_Literatur = ufloat(9.648455,0.000027) * 10 ** 4

print(F_anode,F_kathode, F_Literatur)
```
