```python
from uncertainties import ufloat
import numpy as np
from uncertainties import unumpy as unp
Skalenteilung = ufloat(5,0.13) * 10 ** (-5)
n_Skt = 10
t_sink = 11.07
t_steig = 22.180

v_steig = Skalenteilung * n_Skt / t_steig
v_sink = Skalenteilung * n_Skt / t_sink

eta_0 = 1.81 * 10**(-5)
rho_öl = ufloat(871,2) # TODO temperatur fehler
rho_luft = 1.29
rho = rho_öl - rho_luft
g = 9.81
r_0 = unp.sqrt(9 * eta_0 / (2 * rho * g) * v_sink)
b = 7.78 * 10 ** (-3)
p_Luft = 1000.2 * 100 # in Pascal
eta = eta_0 / (1 + b / (r_0 * p_Luft))
d = ufloat(6, 0.05) * 10**(-3)
U = ufloat(500,3)
q = (v_sink + v_steig) * unp.sqrt(9/2 * v_sink/rho * eta**3 / g) * 6 * np.pi * d / U


```
