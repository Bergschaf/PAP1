```python
Schirm_2 = ufloat(898,1)
Gegenstand_2 = ufloat(200,1)
Linse_pos1 = unp.uarray([356.5,357.5,356.6],[1]*3)
Linse_pos2  = unp.uarray([742.5,743.5,743.5],[1]*3)

Linse_pos1_rot = unp.uarray([359,360,359],[1]*3)
Linse_pos2_rot = unp.uarray([740.5,740.5,741.5],[1]*3)

Linse_pos1_blau = unp.uarray([355,354,354.5],[1]*3)
Linse_pos2_blau = unp.uarray([745,744.5,746.5],[1]*3)

f_Bessel =((Schirm_2-Gegenstand_2)**2 - (Linse_pos1 - Linse_pos2)**2)/(4 * abs(Schirm_2 - Gegenstand_2)) 
f_Bessel_rot =((Schirm_2-Gegenstand_2)**2 - (Linse_pos1_rot - Linse_pos2_rot)**2)/(4 * abs(Schirm_2-Gegenstand_2)) 
f_Bessel_blau =((Schirm_2-Gegenstand_2)**2 - (Linse_pos1_blau - Linse_pos2_blau)**2)/(4 * abs(Schirm_2-Gegenstand_2)) 


#print(f_Bessel)
#print(f_Bessel_rot)
#print(f_Bessel_blau)
print(mittel(f_Bessel), mittel(f_Bessel_rot), mittel(f_Bessel_blau))
```
