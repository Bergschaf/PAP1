```python
from uncertainties import ufloat
from uncertainties import unumpy as unp
from matplotlib import pyplot as plt
import scipy
import numpy as np
```

```python
def sigma(a,b):
    return (a.nominal_value -b.nominal_value)/unp.sqrt((b.std_dev)**2 + (a.std_dev)**2)
```

```python
lam = unp.uarray([408,429,486,640],[3,3,4,9]) * 1e-9
#m = np.array([3,4,5,6])
m = np.array([6,5,4,3])

#print(m)
R_lit = ufloat(1.0973731568157*10**7, 0.000012)
R = 1/(lam * ((1/4) - (1/(m**2))))
print(R)
R_mean = np.mean(R)
print(R_mean)
R_mean = ufloat(1.109, 0.005)*10**7
sigma(R_mean, R_lit)
```

```python
y = np.array([55.8,59.68,20.87,8.1,0.24])
x = np.array([1.5,3,6,12,24])
print(np.e)
def exponential(x, a, b):
    return (a * np.e**(b * x))

value, error = scipy.optimize.curve_fit(exponential, x, y)
plt.plot(x,y)
plt.plot(x,exponential(x,value[0],value[1]))
print(value)
```
